This notebook demonstrates the intermediate transformations, heuristic join logic, and statistical analysis linking CI/CD deployments to ServiceNow incidents.

```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt

# Connect to SQLite analytical database
conn = sqlite3.connect('../data/release_risk.db')
df = pd.read_sql_query('SELECT * FROM deployment_outcomes', conn)
df.head()
```

```python
# Outcome Distribution
print(df['outcome'].value_counts())
print('\nInstability Rate:', round(df['is_instability'].mean() * 100, 2), '%')
```

```python
# Risk by Day of Week
df.groupby('day_of_week')[['is_instability', 'has_rollback']].mean().sort_values(by='is_instability', ascending=False)
```
